```python
from nltk.tokenize import word_tokenize

# Cümleyi sadece tokenleştirip stopwords çıkarma
def preprocess_sentence(sentence):
    tokens = word_tokenize(sentence)  # Cümleyi kelimelere ayır
    # Sadece harf olan kelimeleri al ve stopword'leri çıkar
    filtered_tokens = [token.lower() for token in tokens if token.isalpha() and token.lower() not in stop_words]
    
    return filtered_tokens
```

```python
import json
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

businesses = []
with open("yelp_academic_dataset_business.json", "r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        if i >= 5:  # sadece ilk 5 işletme örnek olsun
            break
        businesses.append(json.loads(line))

# Örnek: ilk işletmenin adı + kategorilerini birleştir
text = businesses[0].get("name", "") + ". " + businesses[0].get("categories", "")

# Cümlelere ayır
sentences = sent_tokenize(text)

filtered_sentences = []
for sentence in sentences:
    tokens = word_tokenize(sentence)
    filtered_tokens = [
        token.lower()
        for token in tokens
        if token.isalpha() and token.lower() not in stop_words
    ]
    filtered_sentences.append(filtered_tokens)

print(filtered_sentences[:10])  # ilk 10 cümle için stopwords temizlenmiş tokenler
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()   # Lemmatizeri başlat
tokenized_corpus_lemmatized = []   # lemma edilmiş cümleleri saklamak için boş liste

for filtered_tokens in filtered_sentences:
    lemmatized_tokens = []
    for token in filtered_tokens:
        lemma = lemmatizer.lemmatize(token)   # Tokenleri tek tek lemma et
        lemmatized_tokens.append(lemma)
    tokenized_corpus_lemmatized.append(lemmatized_tokens)

print(tokenized_corpus_lemmatized[:10])  # ilk 10 cümleyi yazdır
```

```python
import csv

# Lemmatize edilmiş cümleleri bir CSV dosyasına kaydet
with open("lemmatized_sentences.csv", mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    for tokens in tokenized_corpus_lemmatized:
        writer.writerow([' '.join(tokens)])  # her cümle tek satır
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()  # Stemmer'i başlat
tokenized_corpus_stemmed = []   # Stem edilmiş cümleleri saklamak için boş liste

for filtered_tokens in filtered_sentences:
    stemmed_tokens = []
    for token in filtered_tokens:
        stem = stemmer.stem(token)     # Tokenleri stem et
        stemmed_tokens.append(stem)
    tokenized_corpus_stemmed.append(stemmed_tokens)

print(tokenized_corpus_stemmed[:10])  # ilk 10 cümleyi yazdır
```

```python
import csv

# Stem edilmiş cümleleri bir CSV dosyasına kaydet
with open("stemmed_sentences.csv", mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    for tokens in tokenized_corpus_stemmed:
        writer.writerow([' '.join(tokens)])  # her cümle tek satır
```
