```python
import json
from nltk.tokenize import word_tokenize

# Yelp Business Dataset'ten işletmeleri oku
businesses = []
with open("yelp_academic_dataset_business.json", "r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        if i >= 5:  # sadece ilk 5 işletme örnek olsun
            break
        businesses.append(json.loads(line))

# İşletme adı + kategorilerden kelimeler al
kelimeler = []
for biz in businesses:
    text = biz.get("name", "") + " " + biz.get("categories", "")
    tokens = word_tokenize(text)
    kelimeler.extend(tokens)

# İlk 50 kelimeyi göster
print(kelimeler[:50])
```

```python
from nltk.stem import WordNetLemmatizer

# Lemmatizer'ı başlat
lemmatizer = WordNetLemmatizer()

# Lemmatization uygula
lemmatized_words = [lemmatizer.lemmatize(word.lower()) for word in kelimeler]

print(lemmatized_words[:50])  # ilk 50 kelimenin lemma hali
```

```python
from nltk.stem import PorterStemmer

# Stemmer'i başlat
stemmer = PorterStemmer()

# Stemming uygula
stemmed_words = [stemmer.stem(word.lower()) for word in kelimeler]

print(stemmed_words[:50])  # ilk 50 kelimenin stem hali
```

```python
import nltk
nltk.download('wordnet')
from nltk.corpus import wordnet as wn

# Örnek kelime
kelime = "play"

# Synset'leri al
synsets = wn.synsets(kelime)

print(f"'{kelime}' için synset sayısı:", len(synsets))
for s in synsets[:5]:  # ilk 5 synseti gösterelim
    print("Synset:", s.name())
    print("Tanım:", s.definition())
    print("Örnekler:", s.examples())
    print("Üst kavram (hypernym):", s.hypernyms())
    print("Alt kavram (hyponym):", s.hyponyms())
    print("---")
```

```python
from nltk.corpus import wordnet as wn

happy_senses = wn.synsets('happy')
for sense in happy_senses:
    synonyms = sense.lemmas()
    print(f"Synset: {sense.name()} - {sense.definition()}")
    for lemma in synonyms:
        print("  Synonym:", lemma.name())
    print("---")
```

```python
from nltk.corpus import wordnet as wn

hot_senses = wn.synsets('hot')
for sense in hot_senses:
    antonyms = [lemma.antonyms() for lemma in sense.lemmas() if lemma.antonyms()]
    for antonym in antonyms:
        print(f"Synset: {sense.name()} - {sense.definition()}")
        print("  Antonym:", antonym[0].name())
    print("---")
```

```python
from nltk.corpus import wordnet as wn

dog_senses = wn.synsets('dog')
for sense in dog_senses:
    hypernyms = sense.hypernyms()
    print(f"Synset: {sense.name()} - {sense.definition()}")
    for hypernym in hypernyms:
        print("  Hypernym:", hypernym.name(), "-", hypernym.definition())
    print("---")
```

```python
from nltk.corpus import wordnet as wn

car_senses = wn.synsets('car')
for sense in car_senses:
    meronyms = sense.part_meronyms()
    print(f"Synset: {sense.name()} - {sense.definition()}")
    for meronym in meronyms:
        print("  Meronym:", meronym.name(), "-", meronym.definition())
    print("---")
```
